### Différence entre `t()` et `transpose(dim0, dim1)`

- **`t()`** permet de permuter les **lignes et les colonnes** d’une **matrice 2D uniquement**.  
  Il ne prend **aucun paramètre** et échange automatiquement les dimensions `0` et `1`.

- **`transpose(dim0, dim1)`** permet de permuter **deux dimensions quelconques** d’un tenseur.  
  La dimension `0` correspond au **premier élément du `shape`** (par exemple le nombre de lignes) et la dimension `1` au **second** (les colonnes dans le cas d’une matrice 2D).  
  Cette méthode est **plus générale** et s’adapte aux tenseurs de **dimension supérieure ou égale à 2**.

### Résumé

| Fonction | Dimensions supportées | Particularité |
|--------|----------------------|--------------|
| `t()` | 2D uniquement | Raccourci pour la transposition matricielle |
| `transpose()` | ≥ 2D | Permutation explicite des dimensions |

```python
x=torch.rand(3,1)
y=torch.rand(1,4)

print(f"x: {x} \n\n y: {y}\n")

add_x_y=x+y

print(f"Somme : {add_x_y} \n\n Taille : {add_x_y.shape}")
```

### Explication du broadcasting

- On remarque que **chaque ligne de `x` est ajoutée à toutes les colonnes de `y`**.

- **Règles du broadcasting :**
  1. Comparer les dimensions **de droite à gauche**.
  2. Deux dimensions sont compatibles si elles sont **égales** ou si **l’une vaut 1**.
  3. Les dimensions de taille 1 sont **étendues virtuellement** pour correspondre à l’autre.

- Exemple avec `x` de taille `(3,1)` et `y` de taille `(1,4)` :
  - `x` : (3,1) → la dimension 1 est étendue à 4 → (3,4)  
  - `y` : (1,4) → la dimension 0 est étendue à 3 → (3,4)  

- L’addition peut alors se faire **élément par élément** et le résultat a pour taille `(3,4)`.


```python
moy=add_x_y.mean()
som=add_x_y.sum()
var=add_x_y.var()

print(f" Moyenne : {moy} \n\n Somme : {som} \n\n Variance : {var}")
```

```python
moy_by_dim_0=add_x_y.mean(dim=0)
moy_by_dim_1=add_x_y.mean(dim=1)

print(f" Moyenne sur la dim 0 : {moy_by_dim_0} \n\n Moyenne sur la dim 1 : {moy_by_dim_1} \n\n")
```

### Moyennes et moyenne globale en PyTorch

- **Moyenne par colonne (`dim=0`)** : calcule la moyenne de chaque colonne en regroupant toutes les lignes correspondantes.  
- **Moyenne par ligne (`dim=1`)** : calcule la moyenne de chaque ligne en regroupant toutes les colonnes correspondantes.  

- **Moyenne globale (`torch.mean`)** : calcule la somme de **tous les éléments du tenseur** et divise par le **nombre total d’éléments**.  
  Elle donne donc la moyenne de **tous les éléments du tenseur**, sans distinction de lignes ou de colonnes.


### Exercice 5

```python
A=torch.rand(2,3)
B=torch.rand(2,3)
A_cat_B_0=torch.cat((A,B),dim=0)
A_cat_B_1=torch.cat((A,B),dim=1)

print(f"A concat B en dim=0 (ligne) \n\n {A_cat_B_0}\n\n Dimension : {A_cat_B_0.shape}\n")
print(f"A concat B en dim=1 (colone) \n\n {A_cat_B_1}\n\n Dimension : {A_cat_B_1.shape}")


```

```python
C=torch.stack((A,B))
print(f"Stack A et B : \n\n {C} \n\n Dimension : {C.shape}")
```

La **concaténation** consiste à ajouter soit les **lignes**, soit les **colonnes** de tenseurs existantes, en les assemblant le long d’un **axe déjà existant**. Les dimensions doivent être **compatibles sur tous les autres axes**.

Le **stack**, en revanche, crée un **nouveau tenseur** en introduisant une **nouvelle dimension**. Le tenseur résultant est alors composé de **plusieurs sous-tenseurs**, par exemple **deux sous-tenseurs** : l’un contenant les valeurs du tenseur **A** et l’autre celles du tenseur **B**. Les tenseurs empilés doivent avoir **exactement la même forme**.


```python
A=torch.chunk(C,2)[0]
B=torch.chunk(C,2)[1]
print(f"Sous-tenseurs : \n\n {A} \n\n {B}\n\n")
```

### Exercice 7

```python
x=torch.rand(2,2,requires_grad=True)
print(f"x : \n\n {x}")
```

```python
y = (x**2 + 2*x + 1).sum()

print(f"y : {y}")

y.backward()

print(f"dy/dx ou x.grad : \n{x.grad}")
```

```python
x.grad.zero_()
y = (x**2 + 2*x + 1).sum()
print(f"y : {y}")
y.backward()

print(f"x.grad après zero_() et backward: {x.grad}")
```
